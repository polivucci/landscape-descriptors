```python
import sys
import torch
MODULEPATH="../../"
sys.path.append(MODULEPATH)
sys.path.append('../')

from critical_points.torch_critical_point_finder import find_critical_points_torch, save_critical_points_to_csv
from test_schwefel.schwefel_function_nnmodule import Schwefel4D, schwefel_loss_4d  # your existing module
```

```python
# Set parameters
BOUNDS = (0.0, 1.0)
DIM = 4
ATTEMPTS = 64
MIN_DISTANCE = 0.01
```

```python
# Run search
minima, maxima, saddles = find_critical_points_torch(
    model_class=Schwefel4D,
    loss_func=schwefel_loss_4d,
    bounds=BOUNDS,
    dimension=DIM,
    num_attempts=ATTEMPTS,
    min_distance=MIN_DISTANCE,
    device="cpu",
    lr=0.01,
    steps=400
)
```

```python
# Save to CSV
results_df = save_critical_points_to_csv(minima, saddles, dimension=DIM, filename="critical_points_schwefel_torch3D.csv")
display(results_df.head())
```
