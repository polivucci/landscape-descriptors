```python
import sys
import torch
MODULEPATH="../../"
sys.path.append(MODULEPATH)
sys.path.append('../')

from critical_points.torch_critical_point_finder import find_critical_points_torch, save_critical_points_to_csv
from test_schwefel.schwefel_function_nnmodule import Schwefel3D, schwefel_loss  # your existing module
import pandas as pd
```

```python
model_builder = lambda p: Schwefel3D(*p)
```

```python
# Set parameters
BOUNDS = (0.0, 1.0)
DIM = 3
ATTEMPTS = 128
MIN_DISTANCE = 0.01
```

```python
try:
    resume_df = pd.read_csv("critical_points_schwefel_torch3d.csv")
except FileNotFoundError:
    resume_df = None
```

```python
# Run search
minima, maxima, saddles = find_critical_points_torch(
    model_builder=model_builder,
    loss_func=schwefel_loss,
    input = [],             # placeholder model input for compatibility (Schwefel model has no model input)
    bounds=BOUNDS,
    dimension=DIM,
    num_attempts=ATTEMPTS,
    min_distance=MIN_DISTANCE,
    device="cpu",
    lr=5e-2,
    max_iter=5_000,
    resume_df=resume_df,
    seed=1101
  
)
```

```python
# Save to CSV
results_df = save_critical_points_to_csv(minima, maxima, saddles, dimension=DIM, filename="critical_points_schwefel_torch3D.csv")
display(results_df.head())
```
